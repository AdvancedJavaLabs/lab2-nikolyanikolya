```kotlin
import kotlin.io.path.Path

val timeMetricsDf = DataFrame.readJson("graphics/time-metrics.json")
val dataMetricsDf = DataFrame.readJson("graphics/data-metrics.json")
timeMetricsDf
```

```kotlin
timeMetricsDf.plot {
  line {
    x(count)
    y(time)
  }

  layout.title = "execution time (ms) dependency from workers count"
  layout.size = 1000 to 450
}
```

```kotlin
dataMetricsDf.plot {
  line {
    x(size)
    y(time)
  }

  layout.title = "execution time (ms) dependency from file size in MB"
  layout.size = 1000 to 450
}
```
